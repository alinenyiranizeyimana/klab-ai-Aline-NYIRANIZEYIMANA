```python
def to_float(value, default=0.0):
    try:
        return float(value)
    except (TypeError, ValueError):
        return default

print(to_float("24"))
print(to_float("None"))
print(to_float(None))
```

## 3. Three  functions

`normalise()`, `summarise_scores()` and `safe_divide()`.  

```python
def normalise(numbers):
    """Convert a list of numbers to values between 0 and 1."""

    smallest = min(numbers)
    largest = max(numbers)
    difference = largest - smallest

    if difference == 0:
        return [0 for number in numbers]

    scaled_numbers = []

    for number in numbers:
        scaled_numbers.append((number - smallest) / difference)

    return scaled_numbers


student_marks = [45, 60, 75, 90]
print(normalise(student_marks))
```

```python
def summarise_scores(marks):
    """Display simple information about a group of marks."""

    if len(marks) == 0:
        return "No marks available"

    total_marks = sum(marks)
    average_mark = total_marks / len(marks)

    summary = {
        "number_of_marks": len(marks),
        "lowest_mark": min(marks),
        "highest_mark": max(marks),
        "average_mark": average_mark
    }

    return summary


class_marks = [58, 67, 74, 82, 91]
print(summarise_scores(class_marks))
```

```python
def safe_divide(first_number, second_number, fallback=0):
    """Divide two values safely when the second value may be zero."""

    if second_number != 0:
        answer = first_number / second_number
    else:
        answer = fallback

    return answer


print(safe_divide(24, 6))
print(safe_divide(15, 0))
```
